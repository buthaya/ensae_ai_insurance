The `max_depth` parameter in a decision tree or a tree-based ensemble method like XGBoost controls the maximum depth of the tree. It is an important hyperparameter that can significantly affect the performance of the model. 

* **Underfitting**: If the `max_depth` is set too low, the model may not capture the complexity of the data, leading to underfitting. This means the model performs poorly on both the training and test datasets.

* **Overfitting**: If the `max_depth` is set too high, the model may capture too much noise from the training data, leading to overfitting. This means the model performs well on the training dataset but poorly on the test dataset.

* **Optimal `max_depth`**: Finding the optimal `max_depth` involves using cross-validation to evaluate the model's performance on different subsets of the data. By doing this, we can select the `max_depth` that provides the best balance between bias and variance, leading to better generalization on unseen data.

### 4. Use cross-validation to find the best `subsample` parameter

The `subsample` parameter in XGBoost specifies the fraction of the training data that is randomly sampled to grow each tree. 
This technique helps prevent overfitting by introducing randomness into the model training process. 
By using only a subset of the data for each tree, the model becomes less sensitive to noise in the training data and can generalize better to unseen data. 
The `subsample` parameter typically takes values between 0.1 and 1.0, where 1.0 means using all the training data for each tree. Lower values can lead to more robust models but may also increase the variance of the model's predictions.

In this exercise, you will use cross-validation to determine the optimal `subsample` parameter for an XGBoost model on the digits dataset.
